```python
q3=data['price'].quantile(0.75)
q1=data['price'].quantile(0.25)

iqr = q3 - q1

lb = q1 - 1.5 * iqr
ub = q3 + 1.5 * iqr

outliers = data[(data['price'] < lb) | (data['price'] > ub)]
outliers
```

```python
sns.boxplot(x=data['price'])
plt.title("Price Outlier Detection")
plt.show()
```
