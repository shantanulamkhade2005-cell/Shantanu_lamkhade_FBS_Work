```python
emp=pd.DataFrame({'Id':[105,106,107,108],
                'Name':['E','F','G','H'],
                 'Age':[23,45,21,21],
                 'DID':[1,3,2,5]})
```

```python
dept=pd.DataFrame({'DID':[1,2,3,4],
                   'Dname':['It','DS','DA','dev']})
```

```python
pd.merge(emp,dept)
```

```python
dept2=pd.DataFrame({'DI':[1,2,3,4],
                   'Dname':['It','DS','DA','dev']})
```

```python
pd.merge(emp,dept2,left_on='DID',right_on='DI')
```

```python
pd.merge(emp,dept,how='left')
```

```python
pd.merge(emp,dept,how='right')
```

```python
pd.merge(emp,dept,how='inner')
```

```python
pd.merge(emp,dept,how='outer')
```
