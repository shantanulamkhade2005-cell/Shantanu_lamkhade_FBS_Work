```python
plt.figure(figsize=(7,3))
plt.scatter(data[data['Species']=='Setosa']['Sepal_Length'],data[data['Species']=='Setosa']['Sepal_Width'],color='orange',label='Setosa')
plt.scatter(data[data['Species']=='Versicolor']['Sepal_Length'],data[data['Species']=='Versicolor']['Sepal_Width'],color='red',label='Versicolor')
plt.scatter(data[data['Species']=='Virginica']['Sepal_Length'],data[data['Species']=='Virginica']['Sepal_Width'],color='Blue',label='Virginica')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal width')
plt.legend()
plt.show()
```
