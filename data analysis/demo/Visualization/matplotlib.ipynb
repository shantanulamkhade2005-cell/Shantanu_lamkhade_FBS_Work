####   Bar plot
1. comprasing the category
2. showing frequency
3. product sals frequency
4. marks compare

```python
Product=['Tea','Coffee','maggie','juice']
summer=[40,90,75,80]
```

```python
plt.bar(Product,summer,color=['orange','blue','red','black'])
plt.title('Sales in summer')
plt.xlabel('Product')
plt.ylabel('sales')
plt.show()
```

```python
import numpy as np
age=np.random.randint(1,100,50)
bins=[0,20,40,60,80,100]  # use for range of x axis
plt.hist(age,bins=bins)
plt.xlabel('age')
plt.ylabel('count')
plt.title('Age of People ')
plt.show()
```

#### Lineplot
1. Time series analysis
2. Trend over time
3. Upward/Downword trends
4. Ex= IPL team scoresheet

```python
overs=np.arange(1,21)
team1_runs=np.random.randint(0,37,20)
team2_runs=np.random.randint(0,37,20)
```

```python
team1_score=np.cumsum(team1_runs)
team2_score=np.cumsum(team2_runs) #use for sum of series
```

```python
plt.style.use('dark_background')
plt.plot(overs,team1_score,color='red',label='RCB')
plt.plot(overs,team2_score,color='Blue',label='MI')
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.title('Scoresheet')
plt.legend()
plt.show()
```
