```python
import matplotlib.pyplot as plt
marks=[36,78,90,81]
student=['Ram','sham','sita','gita']
plt.title('Marks of Student')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.plot(student,marks,color='orange',linestyle=':',marker='*')
plt.show()
```

```python
marks=[36,78,90,81]
student=['Ram','sham','sita','gita']
plt.title('Marks of Student')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.bar(student,marks,color='orange',edgecolor='black')
plt.show()
```

```python
marks=[36,78,90,81]
student=['Ram','sham','sita','gita']
plt.title('Marks of Student')
plt.xlabel('Student')
plt.ylabel('Marks')
plt.barh(student,marks,color='orange',edgecolor='blue')
plt.show()
```
